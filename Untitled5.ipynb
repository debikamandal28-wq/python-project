Assignment on Functions

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a / b

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition =", addition(a, b))
print("Subtraction =", subtraction(a, b))
print("Multiplication =", multiplication(a, b))

if b != 0:
    print("Division =", division(a, b))
else:
    print("Division is not possible")
```

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

def celsius_to_kelvin(c):
    return c + 273.15

c = float(input("Enter temperature in Celsius: "))
f = float(input("Enter temperature in Fahrenheit: "))

print("Celsius to Fahrenheit =", celsius_to_fahrenheit(c))
print("Fahrenheit to Celsius =", fahrenheit_to_celsius(f))
print("Celsius to Kelvin =", celsius_to_kelvin(c))
```

```python
def check_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

if check_prime(num):
    print(num, "is a Prime Number")
else:
    print(num, "is not a Prime Number")
```

```python
def string_reverse(s):
    return s[::-1]

text = input("Enter a string: ")

result = string_reverse(text)

print("Reversed string =", result)
```

Assignment on Touple, Set and Dictionary

```python
t = (1, 2, 2, 3, 1, 4, 3)

new_tuple = ()

for i in t:
    if i not in new_tuple:
        new_tuple = new_tuple + (i,)

print("Original Tuple:", t)
print("New Tuple:", new_tuple)
```

```python
students = (("Riya", 85), ("Aman", 92),
            ("Soham", 78), ("Priya", 89))

sorted_students = sorted(students, key=lambda x: x[1], reverse=True)

for name, marks in sorted_students:

    print(name, marks)
```

```python
t = (10, 20, 10, 30, 10, 40, 20)

count = 0

for i in t:
    if i == 10:
        count = count + 1

print("Number 10 occurs", count, "times")
```

```python
t = (15, 8, 25, 4, 30, 12)

largest = t[0]
smallest = t[0]

for i in t:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest element:", largest)
print("Smallest element:", smallest)
```

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

if A.issubset(B):
    print("A is a subset of B")
else:
    print("A is not a subset of B")

if B.issuperset(A):
    print("B is a superset of A")
else:
    print("B is not a superset of A")
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] = result[key] + value
    else:
        result[key] = value

print("Merged Dictionary:", result)
```
