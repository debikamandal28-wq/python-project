```python
t = (1, 2, 2, 3, 1, 4, 3)

new = []

for i in t:
    if i not in new:
        new.append(i)

print(tuple(new))
```

```python
students = (("Riya", 85), ("Aman", 92),
            ("Soham", 78), ("Priya", 89))

s = sorted(students, key=lambda x: x[1], reverse=True)

for i in s:
    print(i[0], i[1])
```

```python
a = []

for i in range(5):
    n = int(input("Enter number: "))
    a.append(n)

t = tuple(a)

total = 0

for i in t:
    total = total + i

print("Tuple:", t)
print("Sum:", total)
print("Average:", total / 5)
```

```python
t = (10, 20, 10, 30, 10, 40, 20)

count = 0

for i in t:
    if i == 10:
        count = count + 1

print("Count:", count)
```
